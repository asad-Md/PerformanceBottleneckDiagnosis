```python
# ============================================================
# 1. RAW COLLECTED DATA
# ============================================================

import os
import numpy as np
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 180)

CSV_PARTS = [
    'perf_metrics1.csv',
    'perf_metrics2.csv'
]

dfs = []

for path in CSV_PARTS:
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"{path} not found. Place the CSV files in the same folder as this notebook."
        )

    part = pd.read_csv(path)

    # Same as the v7 notebook
    part['source_file'] = path
    dfs.append(part)

df_raw = pd.concat(dfs, ignore_index=True)

print("=" * 75)
print("RAW COLLECTED DATA")
print("=" * 75)

print("\nData source:")
print("Local eBPF-based performance monitoring")

print("\nInput files:")
for path, part in zip(CSV_PARTS, dfs):
    print(f"  {path:<25} {len(part):>8,} rows")

print("\nCombined raw dataset:")
print(f"  Rows    : {df_raw.shape[0]:,}")
print(f"  Columns : {df_raw.shape[1]}")
print(f"  Shape   : {df_raw.shape}")

print("\nRaw data preview:")
display(df_raw.head(10))

print("\nRaw column names:")
print(df_raw.columns.tolist())
```

```python
# ============================================================
# 2. DATA PREPROCESSING
# Cleaning + Feature Engineering + Feature Selection
# ============================================================

df = df_raw.copy()

rows_before = len(df)

# ------------------------------------------------------------
# STEP 1 — DATA CLEANING
# Same rules as the v7 pipeline
# ------------------------------------------------------------

df = df[df['pid'] > 0]
df = df[df['ctx_switches'] > 0]
df = df[df['latency_count'] > 0]

rows_after_cleaning = len(df)
rows_removed = rows_before - rows_after_cleaning

# ------------------------------------------------------------
# STEP 2 — FEATURE ENGINEERING
# Same causal features as v7
# ------------------------------------------------------------

df['involuntary_ratio'] = (
    df['involuntary_switches'] /
    df['ctx_switches'].clip(lower=1)
)

df['voluntary_ratio'] = (
    df['voluntary_switches'] /
    df['ctx_switches'].clip(lower=1)
)

df['runtime_per_switch'] = (
    df['total_runtime_ns'] /
    df['ctx_switches'].clip(lower=1)
)

df['read_write_ratio'] = (
    df['read_bytes'] /
    (df['write_bytes'] + 1)
)

df['io_bytes_total'] = (
    df['read_bytes'] +
    df['write_bytes']
)

df['alloc_pressure'] = (
    df['total_alloc_bytes'] /
    df['ctx_switches'].clip(lower=1)
)

df['lock_pressure'] = (
    df['mutex_contentions'] +
    df['rwsem_read_contentions'] +
    df['rwsem_write_contentions']
)

# Log-transformed features
log_cols = [
    'total_runtime_ns',
    'avg_syscall_latency_ns',
    'max_syscall_latency_ns',
    'io_bytes_total',
    'total_alloc_bytes'
]

for col in log_cols:
    if col in df.columns:
        df[f'log_{col}'] = np.log1p(df[col])

# ------------------------------------------------------------
# STEP 3 — LEAKAGE-FREE FEATURE SELECTION
# Same allowlist as v7
# ------------------------------------------------------------

LEAKY_COLS = {
    'avg_stall_ns',
    'stall_ns',
    'max_stall_ns',
    'latency_count',
    'stall_per_switch',
    'log_stall_ns',
    'log_avg_stall_ns',
    'log_max_stall_ns',
    'stall_spike',
    'log_stall_spike'
}

ALLOWED_FEATURES = [
    'ctx_switches',
    'cpu_migrations',
    'involuntary_switches',
    'voluntary_switches',
    'involuntary_ratio',
    'voluntary_ratio',
    'avg_runq_ratio',
    'runtime_per_switch',
    'minor_faults',
    'major_faults',
    'kmalloc_count',
    'kfree_count',
    'total_alloc_bytes',
    'alloc_pressure',
    'read_count',
    'write_count',
    'read_bytes',
    'write_bytes',
    'read_write_ratio',
    'io_bytes_total',
    'mutex_contentions',
    'avg_mutex_wait_ns',
    'rwsem_read_contentions',
    'rwsem_write_contentions',
    'lock_pressure',
    'syscall_count',
    'avg_syscall_latency_ns',
    'futex_count',
    'log_total_runtime_ns',
    'log_avg_syscall_latency_ns',
    'log_max_syscall_latency_ns',
    'log_io_bytes_total',
    'log_total_alloc_bytes'
]

FEATURE_COLS = [
    f for f in ALLOWED_FEATURES
    if f in df.columns
]

# Leakage verification
leakage_check = set(FEATURE_COLS) & LEAKY_COLS

assert len(leakage_check) == 0, (
    f"LEAKAGE DETECTED: {leakage_check}"
)

# Final processed feature matrix
X_processed = df[FEATURE_COLS].fillna(0)

# ------------------------------------------------------------
# DISPLAY — SINGLE PREPROCESSING OUTPUT
# ------------------------------------------------------------

print("=" * 75)
print("DATA PREPROCESSING")
print("=" * 75)

print("\n[1] DATA CLEANING")
print(f"  Records before cleaning : {rows_before:,}")
print(f"  Records after cleaning  : {rows_after_cleaning:,}")
print(f"  Records removed         : {rows_removed:,}")

print("\n[2] FEATURE ENGINEERING")
print(f"  Columns after engineering : {df.shape[1]}")
print(f"  New engineered features   : {len(log_cols) + 7}")

print("\n  Engineered features:")
for feature in [
    'involuntary_ratio',
    'voluntary_ratio',
    'runtime_per_switch',
    'read_write_ratio',
    'io_bytes_total',
    'alloc_pressure',
    'lock_pressure',
    'log_total_runtime_ns',
    'log_avg_syscall_latency_ns',
    'log_max_syscall_latency_ns',
    'log_io_bytes_total',
    'log_total_alloc_bytes'
]:
    if feature in df.columns:
        print(f"    ✓ {feature}")

print("\n[3] FEATURE SELECTION")
print(f"  Candidate features : {len(ALLOWED_FEATURES)}")
print(f"  Final ML features  : {len(FEATURE_COLS)}")

print("\n[4] LEAKAGE CHECK")
print("  ✓ PASS — no target-derived stall features included")

print("\n[5] PROCESSED DATA")
print(f"  Final feature matrix shape : {X_processed.shape}")

print("\nProcessed feature data preview:")
display(X_processed.head(10))
```

```python
# ============================================================
# 3. PROCESSED DATA → MODEL INPUT
# ============================================================

print("=" * 75)
print("PROCESSED DATA → MODEL INPUT")
print("=" * 75)

print("\nFinal model input:")
print(f"  Samples  : {X_processed.shape[0]:,}")
print(f"  Features : {X_processed.shape[1]}")
print(f"  Shape    : {X_processed.shape}")

print("\nFeature names supplied to the ML model:")

for i, feature in enumerate(FEATURE_COLS, 1):
    print(f"  {i:2d}. {feature}")

print("\nModel input preview:")
display(X_processed.head(10))

print("\nData types:")
display(X_processed.dtypes.to_frame("dtype"))

print("\nMissing values:")
print(f"  Total missing values: {X_processed.isnull().sum().sum()}")

print("\n" + "=" * 75)
print("DATA FLOW COMPLETED")
print("=" * 75)

print("""
Local eBPF Data Collection
          ↓
Raw CSV Data
          ↓
Data Preprocessing
  • Cleaning
  • Feature Engineering
  • Feature Selection
  • Leakage Check
          ↓
Processed Feature Matrix
          ↓
MODEL INPUT
""")
```
