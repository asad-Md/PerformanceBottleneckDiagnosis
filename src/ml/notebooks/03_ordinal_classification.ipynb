# Phase 3 - Ordinal Classification

This notebook reads Phase 3 artifacts only. It does not train or regenerate models.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
while ROOT.name != 'PerformanceBottleneckDiagnosis' and ROOT.parent != ROOT:
    ROOT = ROOT.parent

ART = ROOT / 'src' / 'ml' / 'artifacts'
ORD = ART / 'ordinal'
comparison = pd.read_csv(ART / 'ordinal_comparison.csv')
summary = comparison.groupby(['model', 'approach'], as_index=False).mean(numeric_only=True)
summary.sort_values(['macro_f1', 'quadratic_weighted_kappa'], ascending=False)
```

## Metric Comparison

Macro F1 measures nominal class identity, while the ordinal metrics below show whether mistakes are near misses or large severity jumps.

```python
plot_df = summary.sort_values('macro_f1', ascending=True)
labels = plot_df['model'] + ' / ' + plot_df['approach']
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].barh(labels, plot_df['macro_f1'])
axes[0].set_title('Macro F1')
axes[0].set_xlabel('score')
axes[1].barh(labels, plot_df['mean_absolute_error'])
axes[1].set_title('Mean absolute class error')
axes[1].set_xlabel('class distance')
fig.tight_layout()
```

## Best Approach Artifacts

The confusion matrix and error distribution are generated by the experiment runner for the best mean Macro F1 approach.

```python
confusion = pd.read_csv(ORD / 'ordinal_confusion_matrix.csv', index_col=0)
display(confusion)
print((ART / 'ordinal_summary.md').read_text(encoding='utf-8'))
```

## Findings

Use the generated `ordinal_summary.md` recommendation as the source of truth for the best model and approach. A useful ordinal model should improve ordered-error metrics such as mean absolute error, adjacent accuracy, severe error rate, and quadratic weighted kappa, even when nominal Macro F1 is close to the baseline.
